```python
import os
from langchain.chat_models import init_chat_model
os.environ["GEMINI_API_KEY"] = os.getenv("gemini")
```

```python
model=init_chat_model("google_genai:gemini-3.5-flash-lite")
```

```python
from pydantic import BaseModel,Field
class Movie(BaseModel):
    """ A Movie with details"""
    title:str=Field(...,description="The title of the movie")
    year:int=Field(...,description="The year the movie was released")
    genre:str=Field(...,description="The genre of the movie")
    director:str=Field(...,description="The director of the movie")
    cast:list[str]=Field(...,description="A list of main actors in the movie")
    plot:str=Field(...,description="A brief summary of the movie's plot")
```

```python
model_with_structure=model.with_structured_output(Movie,include_raw=True)
```

```python
model_with_structure.invoke("Tell me about the movie Inception")
```
